## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `print("hello)` | `SyntaxError` | quotes must come in pairs |
| `Print("hello")` | `NameError` | lowercase `print` — Python is case sensitive |
| `print "hello"` | `SyntaxError` | the brackets are required |
| `'it's fine'` | `SyntaxError` | the apostrophe ends the string — use double quotes |
